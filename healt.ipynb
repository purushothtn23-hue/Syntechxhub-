```python
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
import pandas as pd
```

```python
diabetes=pd.read_csv("diabetes.csv")
diabetes.head(20)
```

```python
diabetes.shape
```

```python
diabetes.isna().sum()
```

```python
diabetes.info()
```

****replace a were 0 it replace a nan****

```python
cols = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']
diabetes[cols] = diabetes[cols].replace(0,np.nan)
```

```python
diabetes.isna().sum()
```

****fill null value by a median value****

```python
diabetes.fillna(diabetes.median(),inplace=True)
diabetes.head(15)
```

****countplot for diabetes****

```python
sns.countplot(x="Outcome",data=diabetes)
plt.title("diabetes distribution")
plt.show()
```

****histplot for age distribution****

```python
sns.histplot(diabetes["Age"],bins=20)
plt.title("age distribution")
plt.show()
```

****histoplot for BMI distribution****

```python
sns.histplot(diabetes["BMI"],bins=20)
plt.title("BMI distribution")
plt.show()
```

****heatmap for fiind a correlation****

```python
plt.Figure(figsize=(10,6))
sns.heatmap(
    diabetes.corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("correlation heatmap")
plt.show()
```

****boxplot for Outcome and Gloucose****

```python
sns.boxplot(
    x="Outcome",
    y="Glucose",
    data=diabetes
)
plt.title("Glucose leve by outcome")
plt.show()
```

****boxplot for outcome and BMI****

```python
sns.boxplot(
    x="Outcome",
    y="BMI",
    data=diabetes
)
plt.title("Glucose leve by outcome")
plt.show()
```

```python
print("\n===== INSIGHTS =====")
print("1. Higher glucose levels are strongly associated with diabetes.")
print("2. Higher BMI increases diabetes risk.")
print("3. Older age groups show higher diabetes prevalence.")
print("4. Glucose has the strongest correlation with Outcome.")
print("5. Lifestyle-related factors significantly impact diabetes occurrence.")
```
