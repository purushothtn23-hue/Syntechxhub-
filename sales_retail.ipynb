```python
import pandas as pd 
import matplotlib.pyplot  as plt
import seaborn as sns 
import numpy as np
```

****load a dataset from kaggle "https://www.kaggle.com/datasets/vivek468/superstore-dataset-final?utm_source=chatgpt.com"****

```python
retail_data = pd.read_csv("Sample - Superstore.csv",encoding="cp1252")
retail_data.head()
```

```python
retail_data.isna().sum()
```

```python
retail_data.info()
```

```python
retail_data.shape
```

```python
retail_data["Order Date"]=pd.to_datetime(retail_data["Order Date"])
```

****total revenue****

```python

total_revenue=retail_data["Sales"].sum()
total_revenue
```

**avg order**

```python
avg_order=retail_data["Sales"].mean()
avg_order
```

****total profit****

```python
total_profit=retail_data["Profit"].sum()
total_profit
```

****find a top saleing product and plot****

```python
top_product=retail_data.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
top_product
```

```python
top_product.plot(kind="bar")
plt.show()
```

****groupby the region and sales ,plot****

```python
top_region=retail_data.groupby("Region")["Sales"].sum()
top_region
```

```python
top_region.plot(kind="bar")
plt.show()
```

***make a monthly order and plot them***

```python
monthly_sales=retail_data.groupby(
    retail_data["Order Date"].dt.month
)["Sales"].sum()
monthly_sales
```

```python
monthly_sales.plot(kind="bar")
plt.show()
```

*****boxplot for month and sales*****

```python
retail_data['Month'] = retail_data['Order Date'].dt.month

sns.boxplot(
    x='Month',
    y='Sales',
    data=retail_data
)

plt.show()
```

****A few products generate most of the revenue.****

****Some regions perform significantly better than others.****

****Sales show seasonal trends.****

****Average order value helps measure customer spending****.

****Monthly analysis helps predict future demand.****
